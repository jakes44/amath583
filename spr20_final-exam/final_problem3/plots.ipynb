```python
import matplotlib.pyplot as plt
import csv
import numpy as np
```

```python
x = []
sequential = []
threaded = []
ideal = []
# bytews, process, time
# 8,4,3.338e-05
with open('./grayscale.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        sequential.append(float(row[1]))
        threaded.append(float(row[2]))
        ideal.append(float(row[1])/int(row[0]))


plt.plot(x, sequential, '--bo', label = "Sequential")
plt.plot(x, threaded, '--ro', label = "Threaded")
plt.plot(x, ideal, '--go', label='Idealized Threaded')

# plt.yscale('log')
plt.xlabel('Number of Threads')
plt.ylabel('Time (ms)')
plt.title('Number of Threads vs Time to Compute Grayscale')
plt.legend(loc='lower left')
# plt.yscale('log')
plt.xticks(x)
plt.grid()
plt.show()
```
