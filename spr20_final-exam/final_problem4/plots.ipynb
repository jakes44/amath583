```python
import matplotlib.pyplot as plt
import csv
import numpy as np
```

```python
x = []
threaded = []
ideal = []
# bytews, process, time
# 8,4,3.338e-05
with open('./scaling.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[1]))
        threaded.append(float(row[2]))
        ideal.append(threaded[0]/int(row[0]))

plt.plot(x, threaded, '--ro', label = "Parallel Time to Compute")
plt.plot(x, ideal, '--go', label='Idealized Scaling')

# plt.yscale('log')
plt.xlabel('Number of Processes')
plt.ylabel('Time (s)')
plt.title('Number of Processes vs Time to Compute Ax=B')
plt.legend(loc='upper right')
plt.yscale('log')
# plt.xscale('log')
plt.xticks(x)
plt.grid()
plt.show()
```
