```python
import matplotlib.pyplot as plt
import csv
```

```python
x = []
y = []
with open('./mm_jki_double_0.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y.append(float(row[1]))

plt.plot(x, y, color = 'r', label = "Double: -O0")

x = []
y = []
with open('./mm_jki_float_0.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y.append(float(row[1]))

plt.plot(x, y, color = 'b', label = "Float: -O0")

x = []
y = []
with open('./mm_jki_double_3.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y.append(float(row[1]))

plt.plot(x, y, color = 'g', label = "Double: -O3")

x = []
y = []
with open('./mm_jki_float_3.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y.append(float(row[1]))

plt.plot(x, y, color = 'orange', label = "Float: -O3")
plt.yscale('log')

plt.xlabel('N (Size of Matrix)')
plt.ylabel('FLOPS')
plt.title('N vs FLOPS for mm_jki')
plt.legend(loc='lower right')
plt.grid()
plt.show()
```
