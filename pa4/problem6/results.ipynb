```python
import numpy as np
import matplotlib.pyplot as plt
import csv
```

```python
x = []
y = []
with open('./swapRows.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y.append(float(row[1]))

plt.plot(x, y, '--ro', label = "Row Swap")

x = []
y = []
with open('./swapCols.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y.append(float(row[1]))

plt.plot(x, y, '--bo', label = "Column Swap")

# plt.yscale('log')
plt.xlabel('N (Dimension of Matrix)')
plt.ylabel('Avg Time (s)')
plt.title('Avg Time for Row/Column Swap')
plt.legend(loc='lower right')
plt.yscale('log')
plt.grid()
plt.show()
```
