```python
import numpy as np
import matplotlib.pyplot as plt
import csv
```

```python
x = []
y = []
with open('./write_speed.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y.append(float(row[1]))

plt.plot(x, y, color = 'r', label = "Write Speed")

x = []
y = []
with open('./read_speed.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y.append(float(row[1]))

plt.plot(x, y, color = 'b', label = "Read Speed")

# plt.yscale('log')
plt.xlabel('N (Dimension of Matrix)')
plt.ylabel('Bytes/sec')
plt.title('Read/Write Speed of Matrix File')
plt.legend(loc='lower right')
plt.grid()
plt.show()
```
