```python
import matplotlib.pyplot as plt
import csv
import numpy as np
```

```python
x = []
y_residual = []
y_error = []
# n, residual, normalized error
with open('./lapack_zgesv.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y_residual.append(float(row[1]))
        y_error.append(float(row[2]))

fig, ax1 = plt.subplots()

ax2 = ax1.twinx()
lns1 = ax1.plot(x, y_residual, '--bo', label = "Residual")
lns_2 = ax2.plot(x, y_error, '--ro', label = "Error")
lns = lns1 + lns_2
labs = [l.get_label() for l in lns]
ax1.legend(lns, labs, loc='center right')

ax1.set_xlabel('Matrix Dimension (n)')
ax1.set_ylabel('Residual')
ax2.set_ylabel('Normalized Error')
ax2.set_yscale('log')
ax1.set_yscale('log')
plt.title('LAPACK zgesv Residual and Normalized Error vs Matrix Dimension')

plt.grid()
plt.show()
```
