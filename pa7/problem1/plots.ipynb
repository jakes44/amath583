```python
import matplotlib.pyplot as plt
import csv
import numpy as np
```

```python
x = []
y_daxpy = []
y_dgemm = []
y_dgemv = []
# bytews, process, time
# 8,4,3.338e-05
with open('./openblas_daxpy.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y_daxpy.append(float(row[1]))

with open('./openblas_dgemv.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        y_dgemv.append(float(row[1]))

with open('./openblas_dgemm.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        y_dgemm.append(float(row[1]))

plt.plot(x, y_daxpy, '--bo', label = "daxpy")
plt.plot(x, y_dgemv, '--go', label = "dgemv")
plt.plot(x, y_dgemm, '--yo', label = "dgemm")

# plt.yscale('log')
plt.xlabel('Problem Size (n)')
plt.ylabel('FLOPS')
plt.title('OpenBLAS Performance - Problem Size vs. FLOPS')
plt.legend(loc='upper right')
plt.yscale('log')
plt.grid()
plt.show()
```
