```python
import matplotlib.pyplot as plt
import csv
import numpy as np
```

```python
x = []
y_d2h = []
y_h2d = []
# bytes, bytes_per_sec_h2d, bytes_per_sec_d2h
with open('./device_copy.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y_h2d.append(float(row[1]))
        y_d2h.append(float(row[2]))

plt.plot(x, y_h2d, '--bo', label = "Host to Device")
plt.plot(x, y_d2h, '--ro', label = "Device to Host")

# plt.yscale('log')
plt.xlabel('Bytes (n)')
plt.ylabel('Throughput (bytes/sec)')
plt.title('Device Copy - Problem Size vs. Throughput')
plt.legend(loc='lower right')
plt.yscale('log')
plt.xscale('log')
plt.grid()
plt.show()
```
