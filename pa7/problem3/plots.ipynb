```python
import matplotlib.pyplot as plt
import csv
import numpy as np
```

```python
x = []
y_open = []
y_cuda = []
# bytews, process, time
# 8,4,3.338e-05
with open('./openblas.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y_open.append(float(row[1]))

with open('./cublas.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        y_cuda.append(float(row[1]))

plt.plot(x, y_open, '--bo', label = "OpenBLAS")
plt.plot(x, y_cuda, '--ro', label = "cuBLAS")

# plt.yscale('log')
plt.xlabel('Problem Size (n)')
plt.ylabel('FLOPS')
plt.title('OpenBLAS & cuBLAS - Problem Size vs. FLOPS')
plt.legend(loc='lower right')
plt.yscale('log')
# plt.xscale('log')
plt.grid()
plt.show()
```
