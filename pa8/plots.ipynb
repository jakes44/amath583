```python
import matplotlib.pyplot as plt
import csv
import numpy as np
```

```python
x = []
y_fftw = []
y_cuda = []
# bytews, process, time
# 8,4,3.338e-05
with open('./fftw.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y_fftw.append(float(row[1]))

with open('./cufft.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        y_cuda.append(float(row[1]))

plt.plot(x, y_fftw, '--bo', label = "FFTW")
plt.plot(x, y_cuda, '--ro', label = "cuFFT")

# plt.yscale('log')
plt.xlabel('Problem Size (n)')
plt.ylabel('FLOPS (log scale)')
plt.title('FFTW & cuFFT - Problem Size vs. FLOPS')
plt.legend(loc='center right')
plt.yscale('log')
# plt.xscale('log')
plt.grid()
plt.show()
```
