```python
import matplotlib.pyplot as plt
import csv
```

```python
x = []
y = []
with open('./multFlops.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    
    for row in plots:
        x.append(int(row[0]))
        y.append(float(row[1]))

plt.plot(x, y, '--bo', label = "Flops")

# plt.yscale('log')
plt.xlabel('N')
plt.ylabel('Floating Point Operations / Time (s)')
plt.title('Avg Flops vs N')
plt.legend(loc='lower right')
# plt.yscale('log')
plt.grid()
plt.show()
```
